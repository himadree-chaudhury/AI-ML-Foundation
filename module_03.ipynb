## Types of Deep Learning Models

###  Artificial Neural Networks (ANNs)
Artificial Neural Networks are the foundational models in deep learning, inspired by the structure and function of the human brain. They consist of layers of interconnected nodes (neurons) that process and transmit information. ANNs are used for a wide range of applications, including classification, regression, and pattern recognition. Feedforward neural networks commonly known as Multi-Layer Perceptrons (MLPs), where information flows in one direction from input to output, are a common type of ANN. They can be further enhanced with techniques like dropout and batch normalization to improve performance and prevent overfitting.

### Convolutional Neural Networks (CNNs)
Convolutional Neural Networks are specialized for processing structured grid data, such as images. They use convolutional layers to automatically and adaptively learn spatial hierarchies of features from input data. CNNs are widely used in image recognition, object detection, and computer vision tasks.

### Recurrent Neural Networks (RNNs)
Recurrent Neural Networks are designed for sequential data, such as time series or natural language. They have connections that form directed cycles, allowing them to maintain a memory of previous inputs. RNNs are commonly used in language modeling, speech recognition, and time series forecasting. RNNs can be further enhanced with architectures like Long Short-Term Memory (LSTM) and Gated Recurrent Units (GRUs) to address issues of vanishing gradients and improve performance on longer sequences. 

### Transformer-Based Models
Transformer-based models have revolutionized natural language processing (NLP) and other sequential data tasks. They utilize self-attention mechanisms to capture long-range dependencies in data, allowing for more efficient training and better performance on tasks like language translation, text generation, and question answering. Notable transformer-based models include BERT, GPT, and T5.

### Large Language Models (LLMs)
Large Language Models are a subset of transformer-based models that have been trained on vast amounts of text data. They are capable of understanding and generating human-like text, making them useful for a wide range of applications, including chatbots, content creation, and language translation. LLMs can perform various tasks such as summarization, sentiment analysis, and code generation, demonstrating a high level of language comprehension and generation capabilities.

### Generative Adversarial Networks (GANs)
Generative Adversarial Networks consist of two neural networks, a generator and a discriminator, that compete against each other. The generator creates synthetic data, while the discriminator evaluates its authenticity. GANs are widely used for generating realistic images, videos, and other types of data, as well as for tasks like data augmentation and style transfer.

### Autoencoders
Autoencoders are unsupervised learning models that learn to encode input data into a lower-dimensional representation and then decode it back to the original input. They are commonly used for dimensionality reduction, anomaly detection, and data compression. Variational Autoencoders (VAEs) are a specific type of autoencoder that can generate new data samples by sampling from the latent space.

### Graph Neural Networks (GNNs)
Graph Neural Networks are designed to work with graph-structured data, where nodes represent entities and edges represent relationships between them. GNNs are used in various applications, including social network analysis, recommendation systems, and molecular chemistry. They can capture complex relationships and dependencies in graph data, making them powerful tools for tasks like node classification, link prediction, and graph generation.



#### Intuition of how the Perceptron Algorithm works:
Let a point $(x_1, x_2) = (6, 5)$ and the actual class is 0. The decision boundary is defined by the equation $w_1 \times x_1 + w_2 \times x_2 + b = 0$, where $w_1$ and $w_2$ are the weights and $b$ is the bias. The predicted class is determined by the sign of the expression $w_1 \times x_1 + w_2 \times x_2 + b$. If the predicted class is 1 then $w_1 \times x_1 + w_2 \times x_2 + b \geq 0$, else if the predicted class is 0 then $w_1 \times x_1 + w_2 \times x_2 + b \leq 0$.\
Let the boundary equation is $2x_1 + 3x_2 + 5 = 0$, then the predicted class is 1 because $2 \times 6 + 3 \times 5 + 5 = 37 \geq 0$. Since the actual class is 0, we need to update the weights and bias beacuse we not move the data point. We will update the weights and bias so that the decision boundary will move and the data point will be on the other side of the boundary. That means we will update the weights and bias so that the new boundary equation will be $w_1 \times x_1 + w_2 \times x_2 + b = 0$ where $w_1$ and $w_2$ are the updated weights and $b$ is the updated bias. The update rule for the weights and bias is as follows:
- If the predicted class is 1 and the actual class is 0, then we will update the weights and bias as follows:
  - $w_1 = w_1 - \alpha \times x_1$
  - $w_2 = w_2 - \alpha \times x_2$
  - $b = b - \alpha$
- If the predicted class is 0 and the actual class is 1, then we will update the weights and bias as follows:
  - $w_1 = w_1 + \alpha \times x_1$
  - $w_2 = w_2 + \alpha \times x_2$
  - $b = b + \alpha$
- Combining the above two cases, we can write the update rule as follows:\
  $w_i = w_i + \alpha \times (predicted\_class - actual\_class) \times x_i$\
  If the predicted class is 1 and the actual class is 0, then $predicted\_class - actual\_class = 1 - 0 = 1$, and if the predicted class is 0 and the actual class is 1, then $predicted\_class - actual\_class = 0 - 1 = -1$. 

In this example, since the predicted class is 1 and the actual class is 0, we will update the weights and bias as follows:
- $w_1 = 2 - \alpha \times 6$
- $w_2 = 3 - \alpha \times 5$
- $b = 5 - \alpha$

Let, $\alpha = 0.1$, then the updated weights and bias will be as follows:
- $w_1 = 2 - 0.1 \times 6 = 1.4$
- $w_2 = 3 - 0.1 \times 5 = 2.5$
- $b = 5 - 0.1 = 4.9$

The new boundary equation will be $1.4x_1 + 2.5x_2 + 4.9 = 0$. Now, we will check the predicted class for the point (6, 5) again using the new boundary equation. The predicted class is determined by the sign of the expression $1.4 \times 6 + 2.5 \times 5 + 4.9$. If the predicted class is 1 then $1.4 \times 6 + 2.5 \times 5 + 4.9 \geq 0$, else if the predicted class is 0 then $1.4 \times 6 + 2.5 \times 5 + 4.9 \leq 0$.\
Calculating the expression, we get $1.4 \times 6 + 2.5 \times 5 + 4.9 = 29.9 \geq 0$. Since the predicted class is still 1 and the actual class is 0, we will update the weights and bias again using the same update rule. The updated weights and bias will be as follows:
- $w_1 = 1.4 - 0.1 \times 6 = 0.8$
- $w_2 = 2.5 - 0.1 \times 5 = 2.0$
- $b = 4.9 - 0.1 = 4.8$

Now, the new boundary equation will be $0.8x_1 + 2.0x_2 + 4.8 = 0$. We will check the predicted class for the point (6, 5) again using the new boundary equation. The predicted class is determined by the sign of the expression $0.8 \times 6 + 2.0 \times 5 + 4.8$. If the predicted class is 1 then $0.8 \times 6 + 2.0 \times 5 + 4.8 \geq 0$, else if the predicted class is 0 then $0.8 \times 6 + 2.0 \times 5 + 4.8 \leq 0$.\
Calculating the expression, we get $0.8 \times 6 + 2.0 \times 5 + 4.8 = 25.6 \geq 0$. Since the predicted class is still 1 and the actual class is 0, we will update the weights and bias again using the same update rule. This process will continue until the predicted class is 0 and the actual class is 0, which means the weights and bias have converged and the decision boundary has moved to the correct position.\
Ultimately, the deceision boundary will become $-0.2x_1 + 0.5x_2 + 0.5 = 0$ and the predicted class for the point (6, 5) will be 0 because $-0.2 \times 6 + 0.5 \times 5 + 0.5 = -0.2 \leq 0$.

This way the Perceptron Algorithm creates a decision boundary that separates the data points of different classes. The algorithm iteratively updates the weights and bias based on the predicted class and the actual class of the data points, moving the decision boundary until it correctly classifies all the training data points.

##### Why learning rate is important in the Perceptron Algorithm?
The learning rate, denoted as $\alpha$, is a crucial hyperparameter in the Perceptron Algorithm that controls the step size of the weight and bias updates during training. It determines how much the weights and bias are adjusted in response to the error between the predicted class and the actual class. A learning rate that is too high can cause the algorithm to overshoot the optimal weights and bias, leading to divergence or oscillation in the training process. On the other hand, a learning rate that is too low can result in slow convergence, requiring many iterations to reach the optimal solution. Therefore, choosing an appropriate learning rate is essential for ensuring that the Perceptron Algorithm converges efficiently to a solution that correctly classifies the training data.
