# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

## Write Answer 01:
Gradient descent is an optimization algorithm used to minimize the loss function in machine learning models. The main idea behind gradient descent is to iteratively adjust the model's parameters in the direction of the steepest descent of the loss function. This is achieved by calculating the gradient of the loss function with respect to the model's parameters and updating the parameters in the opposite direction of the gradient.

The learning rate is a hyperparameter that determines the step size at each iteration while moving towards a minimum of the loss function. If the learning rate is too high, the algorithm may overshoot the minimum, leading to divergence or oscillation. On the other hand, if the learning rate is too low, the algorithm may take a very long time to converge, resulting in slow training and potentially getting stuck in local minima. Therefore, choosing an appropriate learning rate is crucial for the success of gradient descent in minimizing the loss function effectively.

# Question 02: [ Marks 10 ]

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Write Answer 02:
Forward propagation is the process by which data moves through an artificial neural network from the input layer to the output layer. The forward propagation process allows the neural network to learn complex patterns in the data by adjusting the weights and biases during training, ultimately enabling it to make accurate predictions on new, unseen data. It involves several steps:
1. Input Layer: The process begins at the input layer, where the raw data is fed into the network. Each neuron in the input layer represents a feature of the data.
2. Hidden Layers: The data then moves to the hidden layers. Each neuron in the hidden layer receives all input from the previous layer, applies a weighted sum of the inputs, and then applies an activation function to introduce non-linearity. This process is repeated for each hidden layer in the network.
3. Output Layer: Finally, the data reaches the output layer, where the neurons produce the final output of the network.

# Question 03: [ Marks 15 ]
Why do we need activation functions in neural networks? What would happen if we only used linear transformations without any activation function?

## Write Answer 03:
Activation functions are essential in neural networks because they introduce non-linearity into the model. Without activation functions, a neural network would only be able to learn linear relationships between the input and output data. As a result, the network would not be able to capture complex patterns or relationships in the data, limiting its ability to solve real-world problems that often involve non-linear relationships. Therefore, activation functions enable neural networks to learn and model complex data effectively. Using only linear transformations without any activation function would result a model being a linear regression, which is insufficient for tasks such as image recognition, natural language processing, and other complex problems that require the ability to learn non-linear relationships.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Write Answer 04:
The sigmoid activation function is a mathematical function that converts any real-valued input to a value between 0 and 1. It is defined as:
```math
\sigma(x) = \frac{1}{1 + e^{-x}}
```
The sigmoid function is commonly used in the output layer of binary classification problems, where the output represents the probability of belonging to a particular class. Setting a threshold (typically 0.5) allows the model to make binary predictions. It is also used in hidden layers of neural networks, although it has been largely replaced by other activation functions like ReLU due to its disadvantages.

Advantages of the sigmoid activation function include:
1. It outputs values in a range between 0 and 1, which can be interpreted as probabilities.
2. It is smooth and differentiable, which allows for gradient-based optimization methods like backpropagation to work effectively.

Disadvantages of the sigmoid activation function include:
1. It can cause the vanishing gradient problem during training, where the gradients become very small for large positive or negative input values, leading to slow convergence or even stopping the training process.
2. It is not zero-centered, which can lead to inefficient training and slower convergence.
3. It can saturate and kill gradients, meaning that neurons can become inactive and stop learning if the input values are too large or too small.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:
The softmax activation function is a mathematical function that converts a vector of real-valued inputs into a probability distribution. It is defined as:
```math
\sigma(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}
```
Where $z_i$ is the input to the $i$-th neuron and $K$ is the total number of classes. The softmax function takes a vector and transforms it into probabilities that sum to 1, making it ideal for multi-class classification problems where we want to predict the probability of each class.

The softmax function is different from the sigmoid function in that sigmoid is used for binary classification problems and outputs a single probability value between 0 and 1, while softmax is used for multi-class classification problems and outputs a probability distribution over multiple classes. In other words, sigmoid is suitable for problems where there are only two classes, while softmax is necessary when there are more than two classes to predict.

# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Write Answer 06:
An Artificial Neural Network (ANN) is inspired by the structure and function of the human brain. It consists of layers of interconnected nodes, called neurons, which process and transmit information. The structure of an ANN typically includes three main types of layers: the input layer, hidden layers, and the output layer.
1. Input Layer: The input layer is the first layer of the ANN and is responsible for receiving the raw data. Each neuron in the input layer represents a feature of the data. For example, if we are working with an image classification task, each neuron in the input layer might represent a pixel value of the image.
2. Hidden Layers: The hidden layers are the intermediate layers between the input and output layers. They are responsible for processing the input data and extracting features. Each neuron in the hidden layer receives all input from the previous layer, applies a weighted sum of the inputs, adds a bias term, and then applies an activation function to introduce non-linearity. This process allows the network to learn complex patterns in the data. For instance, in an image classification task, the hidden layers might learn to detect edges, shapes, and other features that are important for distinguishing between different classes.
3. Output Layer: The output layer is the final layer of the ANN and produces the final output of the network. The number of neurons in the output layer depends on the type of problem being solved. For a binary classification problem, there might be a single neuron that outputs a probability value, while for a multi-class classification problem, there might be multiple neurons that output a probability distribution over the classes.

Weights, biases, and activation functions work together to transform input data into predictions. The weights determine the strength of the connection between neurons, while the biases allow the model to shift the activation function to better fit the data. The activation function introduces non-linearity, enabling the network to learn complex patterns.

For example, consider a simple ANN for a binary classification task with one input layer (2 neurons), one hidden layer (3 neurons), and one output layer (1 neuron). The input layer receives two features of the data, which are then processed by the hidden layer. Each neuron in the hidden layer applies a weighted sum of the inputs, adds a bias, and applies an activation function. The output from the hidden layer is then passed to the output layer, where it is transformed into a probability value using a sigmoid activation function. This probability value indicates the likelihood of the input belonging to a particular class. Through training, the weights and biases are adjusted to minimize the loss function, allowing the ANN to make accurate predictions on new data.
