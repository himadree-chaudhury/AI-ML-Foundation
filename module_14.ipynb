# Pytorch NN Module
The Pytorch NN module is a neural network module designed for natural language processing tasks. It provides a flexible and efficient way to build and train neural networks for various NLP applications.\
It abstracts the complexity of creating and training neural networks by offering pre-built layers,loss functions, activation functions, and other utilities, enabling you to focus on designing and experimenting with model architectures.

### Key Components of torch.nn:

1. Modules (Layers):
   - nn.Module: The base class for all neural network modules. Custom models and layers can be created by subclassing nn.Module and defining the forward method.
   - Common Layers: Includes layers like nn.Linear (fully connected layer), nn.Conv2d (convolutional layer), nn.LSTM (recurrent layer), and many others.
2. Activation Functions: Functions like nn.ReLU, nn.Sigmoid, and nn.Tanh introduce non-linearities to the model, allowing it to learn complex patterns.
3. Loss Functions: Provides loss functions such as nn.CrossEntropyLoss, nn.MSELoss, and nn.NLLLoss to quantify the difference between the model's predictions and the actual targets.
4. Container Modules:
   - nn.Sequential: A sequential container to stack layers in order.
5. Regularization and Dropout: Layers like nn.Dropout and nn.BatchNorm2d help prevent overfitting and improve the model's ability to generalize to new data.

### PyTorch Optimization Details

* torch.optim: A module in PyTorch providing optimization algorithms to update model parameters during training.
* Common Optimizers: Includes Stochastic Gradient Descent (SGD), Adam, RMSprop, and others.
* Efficiency & Features: Handles weight updates and includes learning rate scheduling and weight decay (regularization).
* model.parameters(): A method that retrieves an iterator over all trainable parameters (weights and biases).
* Weights: The weight matrices of layers like nn.Linear, nn.Conv2d, etc.
* Biases: The bias terms of layers (if they exist).
* Optimizer Role: Uses these parameters to compute gradients and update them during training.

#### NN Container Modules
A container module in PyTorch is a special type of module that can hold and manage other modules (layers) in a structured way. It allows to organize layers sequentially or in a more complex architecture without having to manually define the forward pass for each layer. The most common container module is nn.Sequential, which stacks layers in the order they are added.

```python   
import torch
import torch.nn as nn

class Model(nn.Module):
    def __init__(self, feature_size):
        super(Model, self).__init__()

        # Defining a sequential container to hold 4 layers: linear(input), activation, linear, activation
        self.network = nn.Sequential(
            nn.Linear(feature_size, 3), #index 0, with input size = feature_size and output size = 3
            nn.ReLU() ,   #index 1, with no parameters
            nn.Linear(3,1),  #index 2, with input size = 3 and output size = 1
            nn.Sigmoid()  #index 3, activation function with no parameters
        )

    def forward(self, X):
        return self.network(X)
```

````python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
````

````python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
````

````python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)

X_train_tensor = torch.from_numpy(X_train).float()
X_test_tensor = torch.from_numpy(X_test).float()
y_train_tensor = torch.from_numpy(y_train).float()
y_test_tensor = torch.from_numpy(y_test).float()
````

````python
import torch.nn as nn

class NeuralNetwork(nn.Module):
    def __init__(self, feature_numbers):
        super().__init__()      # Inherit nn.Module class

        self.linear = nn.Linear(feature_numbers, 1) # Linear layer with input, output size = feature_numbers and output size = 1
        self.sigmoid = nn.Sigmoid() # Sigmoid activation function for binary classification

    def forward(self, X):
        output = self.linear(X) # Linear transformation
        y_pred = self.sigmoid(output) # Apply sigmoid activation to get probabilities between 0 and 1
        return y_pred
    
learning_rate = 0.1
epochs = 100
loss_function = nn.BCELoss()    # Binary Cross Entropy Loss for binary classification

model = NeuralNetwork(X_train_tensor.shape[1])
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

for epoch in range(epochs):
    y_pred = model.forward(X_train_tensor)
    loss = loss_function(y_pred, y_train_tensor.reshape(-1, 1))

    optimizer.zero_grad()  # Clear gradients before backpropagation

    loss.backward()
    
    optimizer.step() # Update weights and biases based on gradients
    
    print(f'Epoch {epoch+1}/{epochs}, Loss: {loss.item()}')
````

````python
with torch.no_grad():
    y_pred_test = model.forward(X_test_tensor)
    y_pred_labels = (y_pred_test > 0.5).float()
    accuracy = (y_pred_labels.squeeze() == y_test_tensor).float().mean()
    print(f'Test Accuracy: {accuracy.item() * 100:.2f}%')
````
