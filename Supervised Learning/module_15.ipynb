# Support Vector Machine (SVM)

Support Vector Machine (SVM) is a powerful supervised machine learning algorithm used for both classification and regression tasks. It works by finding the optimal hyperplane that separates data points of different classes in a high-dimensional space. SVM aims to maximize the margin between the closest data points (support vectors) of each class and the hyperplane, leading to better generalization on unseen data.

## Support Vectors
Support vectors are the data points that are closest to the decision or class boundary. These points are critical because they directly influence the position and orientation of the hyperplane. The SVM algorithm focuses on these support vectors to determine the optimal hyperplane that maximizes the margin. These points are often called boundary points because they lie on the edge of the margin. For stable boundary and outlier handling, SVM relies on these support vectors rather than the entire dataset.

```math
Support Vectors = \{x_i | y_i (w · x_i + b) = 1\}
```
where, `xi` are the support vectors, `yi` are their corresponding class labels(+1/-1), `w` is the weight vector, and `b` is the bias term.

## Hyperplane
A hyperplane is a decision boundary that separates different classes in the feature space. In two dimensions, it is a line; in three dimensions, it is a plane; and in higher dimensions, it is referred to as a hyperplane.

## Margin Intuition
The margin is the distance from the decision boundary to the nearest data points on either side.\
`Bigger margin = Better generalization`\
A wide margin means the model is confident about its classifications and less likely to misclassify new, unseen data.
- Hard Margin: No misclassifications allowed, which can lead to overfitting.
- Soft Margin: Allows some misclassifications to improve generalization.

```math
w · x + b = 0
```

Svm construts two margin boundaries:

```math
w · x + b = 1; w · x + b = -1
```
where, `w` is the weight vector, `x` is the input feature vector, and `b` is the bias term.

```math
Margin = \frac{2}{||w||}
```

Example: Let, the boundary is -2x+6=0\
Here, w=-2, b=6\
So, Margin = 2/|-2|=1

## Hinge Loss
Hinge loss is a loss function used in SVMs to measure the error in classification tasks. It penalizes misclassified points and those that are within the margin, encouraging the model to find a decision boundary that maximizes the margin while minimizing classification errors.

```math
L_{hinge}(x_i) = \max(0, 1 - y_i (w \cdot x_i + b))
```
where, `Lhinge(xi)` is the hinge loss for a data point `xi`, `yi` is the true class label (+1 or -1), `w` is the weight vector, `b` is the bias term.
1. If a data point is correctly classified and far from the margin, the hinge loss is zero.
2. If a data point is correctly classified but close to the margin, loss is positive.
3. If a data point is missclassified, loss becomes larger.

Example-1: Let, X=[3,2], y=1, w=[1,1], b=-4\
Decesion score = w·X + b = 1*3 + 1*2 - 4 = 1\
Hinge Loss = max(0, 1 - 1*1) = 0\
That means, the point is correctly classified and outside the margin.

Example-2: Let, X=[0,1], y=1, w=[2,1], b=-3\
Decesion score = w·X + b = 2*0 + 1*1 - 3 = -2\
Hinge Loss = max(0, 1 - 1*(-2)) = 3\
That means, the point is misclassified and inside the margin.

## Kernels
Kernels are functions used in SVMs to transform data into a higher-dimensional space, allowing the algorithm to find a linear separation in cases where the data is not linearly separable in its original space. Kernels enable SVMs to efficiently compute the inner products of data points in this higher-dimensional space without explicitly performing the transformation, which can be computationally expensive.

### Kernel Trick
The kernel trick is a technique used in SVMs to efficiently compute the inner products of data points in a higher-dimensional space without explicitly transforming the data into that space. This is achieved by using kernel functions that calculate the inner product directly in the original feature space. The kernel trick allows SVMs to handle non-linearly separable data by implicitly mapping it to a higher-dimensional space where a linear separation is possible, all while avoiding the computational complexity of working in that high-dimensional space.

Common kernel functions include:
- Linear Kernel: Suitable for linearly separable data. Defined as k(x, y) = xT · y, where x and y are input vectors, T denotes the transpose operation.
- Polynomial Kernel: Captures non-linear relationships by considering polynomial combinations of features. Defined as k(x, y) = (xT · y + c)^d, where c is bias and d is the degree of the polynomial. Higher degreee often leads to overfitting.
- Radial Basis Function (RBF) Kernel: Maps data into an infinite-dimensional space, effective for complex decision boundaries. It measures similarity via Euclidean distance, creating localized influence around support vectors. Defined as k(x, y) = exp(-γ ||x - y||^2), where γ (gamma) controls the width of the Gaussian function.
 Key Characteristics of RBF Kernel:
    - Infinite Dimensional: RBF kernel maps data into an infinite-dimensional space, allowing for highly flexible decision boundaries.
    - Localized Influence: The RBF kernel creates a localized influence around each support vector, meaning that points closer to a support vector have a greater impact on the decision boundary.
    - Shape Adaptation: The RBF kernel can adapt to various data distributions, making it suitable for a wide range of applications.

- Sigmoid Kernel: Similar to neural networks, useful for certain types of data.
- Laplacian Kernel: Similar to RBF but uses the L1 norm, making it more robust to outliers.
- rbfpoly Kernel: A combination of RBF and Polynomial kernels, capturing both local and global patterns in the data.
- linrbf Kernel: A hybrid of Linear and RBF kernels, balancing simplicity and complexity in modeling.
- etakernel: A custom kernel that can be designed to capture specific patterns in the data, often based on domain knowledge.
The choice of kernel significantly impacts the performance of the SVM model, and selecting the appropriate kernel often requires experimentation and cross-validation.


### Gamma (γ) Parameter
The gamma (γ) parameter in SVMs, particularly when using the Radial Basis Function (RBF) kernel, controls the influence of individual training examples on the decision boundary. It determines how far the influence of a single training example reaches. A low gamma value means that the influence reaches far, resulting in a smoother decision boundary that may underfit the data. Conversely, a high gamma value means that the influence is limited to nearby points, leading to a more complex decision boundary that may overfit the training data. Therefore, selecting an appropriate gamma value is crucial for balancing bias and variance in the SVM model. Tuning gamma often involves experimentation and cross-validation to find the optimal value for a specific dataset.

### C Parameter
The C parameter in SVMs is a regularization parameter that controls the trade-off between maximizing the margin and minimizing classification errors. A low C value(strict classification) allows for a wider margin, which can lead to better generalization on unseen data but may result in more misclassifications on the training set. Conversely, a high C value(relaxed classification) prioritizes minimizing classification errors on the training set, leading to a narrower margin that may overfit the data. Therefore, selecting an appropriate C value is crucial for balancing bias and variance in the SVM model. Tuning C often involves experimentation and cross-validation to find the optimal value for a specific dataset.

## Support Vector Classifier (SVC)
Support Vector Classifier (SVC) is a specific implementation of the Support Vector Machine (SVM) algorithm designed for classification tasks. SVC aims to find the optimal hyperplane that separates different classes in the feature space while maximizing the margin between the classes. It utilizes support vectors, which are the data points closest to the decision boundary, to determine the position and orientation of the hyperplane. SVC can handle both linearly separable and non-linearly separable data by employing kernel functions to transform the data into a higher-dimensional space. The algorithm also incorporates regularization techniques, such as soft margins, to allow for some misclassifications and improve generalization on unseen data. Overall, SVC is a powerful and versatile classification algorithm widely used in various machine learning applications.

### Why SVM Matters?
- Data efficient: Works well with smaller datasets.
- Strong boundaries: Finds optimal decision boundaries.
- Nonlinear handling: Uses kernels for complex data.
- Simple yet powerful: Effective for various tasks.

### When to Use SVM?
- Medium-sized Datasets
- Clear Separability
- Nonlinear Relationships
- Robustness Required

````python
import numpy as np
import matplotlib.pyplot as plt

# Helper function to plot decision boundary for 2D data
def plot_decision_boundary(model, X, y, title='Decision boundary'):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 300),
        np.linspace(y_min, y_max, 300)
    )
    grid = np.c_[xx.ravel(), yy.ravel()]
    Z = model.predict(grid)
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.2, cmap='bwr')
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr', edgecolors='k', alpha=0.8)
    plt.title(title)
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.show()
````

````python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

cancer = load_breast_cancer()
X_bc = cancer.data[:,:2]        # Use only the first two features
y_bc = cancer.target

X_bc_train, X_bc_test, y_bc_train, y_bc_test = train_test_split(
    X_bc, y_bc, test_size=0.2, random_state=42, stratify=y_bc
)

scaler_bc = StandardScaler()
X_bc_train_scaled = scaler_bc.fit_transform(X_bc_train)
X_bc_test_scaled = scaler_bc.transform(X_bc_test)
````

````python
# Linear kernel on dataset
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

svc_linear = SVC(kernel='linear', C=1.0, random_state=42)
svc_linear.fit(X_bc_train_scaled, y_bc_train)
plot_decision_boundary(svc_linear, X_bc_train_scaled, y_bc_train,
                       title='Linear SVM on Breast Cancer (First 2 Features)')

linear_pred= svc_linear.predict(X_bc_test_scaled)
linear_accuracy = accuracy_score(y_bc_test, linear_pred)
print(f'Accuracy (using only 2 features): {linear_accuracy:.4f}')
````

````python
# Polynomial kernel on dataset
svc_poly = SVC(kernel='poly', degree=3, C=1.0, random_state=42)
svc_poly.fit(X_bc_train_scaled, y_bc_train)
plot_decision_boundary(svc_poly, X_bc_train_scaled, y_bc_train,
                       title='Polynomial SVM on Breast Cancer (First 2 Features)')

poly_pred = svc_poly.predict(X_bc_test_scaled)
poly_accuracy = accuracy_score(y_bc_test, poly_pred)
print(f'Accuracy (using only 2 features): {poly_accuracy:.4f}')
````

````python
# RBF kernel on dataset
svc_rbf = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)
svc_rbf.fit(X_bc_train_scaled, y_bc_train)
plot_decision_boundary(svc_rbf, X_bc_train_scaled, y_bc_train,
                       title='RBF SVM on Breast Cancer (First 2 Features)')

rbf_pred = svc_rbf.predict(X_bc_test_scaled)
rbf_accuracy = accuracy_score(y_bc_test, rbf_pred)
print(f'Accuracy (using only 2 features): {rbf_accuracy:.4f}')
````

````python
# Effect of Hyperparameters (C and gamma) on RBF Kernel SVM
import pandas as pd

def evaluate_svm_c_gamma(X_train, X_test, y_train, y_test, C_values, gamma_values):
    results = []
    for C in C_values:
        for gamma in gamma_values:
            model = SVC(kernel='rbf', C=C, gamma=gamma, random_state=42)
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            acc = accuracy_score(y_test, y_pred)
            results.append({'C': C, 'gamma': gamma, 'accuracy': acc})
    return pd.DataFrame(results)

C_values = [0.1, 1.0, 10.0]
gamma_values = ['scale', 0.01, 0.1, 1.0]

cancer_data = load_breast_cancer()
X_bc_data = cancer_data.data
y_bc_data = cancer_data.target

X_bc_data_train, X_bc_data_test, y_bc_data_train, y_bc_data_test = train_test_split(
    X_bc_data, y_bc_data, test_size=0.2, random_state=42, stratify=y_bc_data
)

scaler_bc_data = StandardScaler()
X_bc_data_train_scaled = scaler_bc_data.fit_transform(X_bc_data_train)
X_bc_data_test_scaled = scaler_bc_data.transform(X_bc_data_test)

df_hyper = evaluate_svm_c_gamma(X_bc_data_train_scaled,X_bc_data_test_scaled, y_bc_data_train, y_bc_data_test,C_values, gamma_values)
df_hyper
````

## Suppor Vector Regression (SVR)
Support Vector Regression (SVR) is an extension of the Support Vector Machine (SVM) algorithm designed for regression tasks. SVR aims to find a function that approximates the relationship between input features and continuous target values while maintaining a balance between model complexity and prediction accuracy. Similar to SVM, SVR utilizes support vectors, which are the data points closest to the decision boundary, to determine the optimal regression function. The algorithm employs a margin of tolerance (epsilon) around the predicted values, allowing for some flexibility in fitting the data. SVR can also utilize kernel functions to handle non-linear relationships between features and target values. Overall, SVR is a powerful regression technique that provides robust predictions while controlling overfitting through regularization.

### The Epsilon-Insensitive Tube
In Support Vector Regression (SVR), the epsilon-insensitive tube is a concept that defines a margin of tolerance around the predicted values, within which no penalty is applied for prediction errors. This means that if the predicted value falls within this tube (i.e., the difference between the predicted and actual value is less than epsilon), the model does not incur any loss. The purpose of the epsilon-insensitive tube is to allow for some flexibility in the regression model, enabling it to focus on capturing the overall trend of the data rather than fitting every single data point exactly. This helps to prevent overfitting and improves the generalization of the model to unseen data. The width of the epsilon-insensitive tube is controlled by the epsilon parameter, which can be adjusted based on the desired level of tolerance for prediction errors.

- The tolerance zone by consrtructing an epsilon tube
- Zero loss within this zone
- Beyond this zone, loss increases linearly

### Handling Violation with Slack Variables
In Support Vector Regression (SVR), slack variables are introduced to handle violations of the epsilon-insensitive tube. When a data point falls outside the epsilon tube, it incurs a loss, and slack variables are used to quantify the extent of this violation. Each slack variable represents the amount by which a data point exceeds the epsilon margin, either above or below the predicted value. The SVR optimization process aims to minimize both the magnitude of these slack variables and the overall complexity of the regression function. By incorporating slack variables, SVR can effectively manage outliers and ensure that the model remains robust while still capturing the underlying patterns in the data. This approach allows for a balance between fitting the data closely and maintaining generalization to unseen data points. 

```math
Minimize: \frac{1}{2} ||w||^2 + C \sum_{i=1}^{n} (\xi_i + \xi_i^*)
```
```math
Costraints:
y_i - (w \cdot x_i + b) \leq \epsilon + \xi_i\\
(w \cdot x_i + b) - y_i \leq \epsilon + \xi_i^*\\
\xi_i, \xi_i^* \geq 0
```
where, `C` is the regularization parameter, `ξi` and `ξi*` are the slack variables for data point `i`, `w` is the weight vector, `b` is the bias term, `xi` is the input feature vector, and `yi` is the actual target value.

Example: Let, for a data point xi, the actual value yi=5, predicted value f(xi)=3, and epsilon=1, w=[1], b=2\
Here, the point violates the epsilon tube since |5-3|=2 > 1\
Slack variable ξi = yi - (w · xi + b) - ε = 5 - 3 - 1 = 1\
This slack variable indicates that the data point is 1 unit outside the epsilon-insensitive tube.\
Minimize = 1/2 * ||[1]||^2 + C * (1 + 0) = 0.5 + C * 1 = 0.5 + C

#### SVR vs Linear Regression
- SVR focuses on fitting within a margin (epsilon tube), while Linear Regression minimizes overall error.
- SVR uses support vectors to define the regression function, while Linear Regression uses all data points.
- SVR can handle non-linear relationships using kernels, while Linear Regression is limited to linear relationships.
- SVR is more robust to outliers due to the epsilon-insensitive tube, while Linear Regression can be sensitive to outliers.
- SVR includes regularization through the C parameter, while Linear Regression typically does not have built-in regularization (unless using variants like Ridge or Lasso regression).

````python
from sklearn.datasets import load_diabetes

diabetes = load_diabetes()
X_diab = diabetes.data
y_diab = diabetes.target

X_diab_train, X_diab_test, y_diab_train, y_diab_test = train_test_split(
    X_diab, y_diab, test_size=0.2, random_state=42
)

scaler_diab = StandardScaler()
X_diab_train_scaled = scaler_diab.fit_transform(X_diab_train)
X_diab_test_scaled = scaler_diab.transform(X_diab_test)
````

````python
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score

svr_diab = SVR(kernel='rbf', C=10.0, epsilon=0.1, gamma='scale')
svr_diab.fit(X_diab_train_scaled, y_diab_train)

y_diab_pred = svr_diab.predict(X_diab_test_scaled)

mse_diab = mean_squared_error(y_diab_test, y_diab_pred)
r2_diab = r2_score(y_diab_test, y_diab_pred)
# Here accuracy is not a suitable metric because this is a regression task which predicts continuous values not discrete classes, like 0 or 1.

print(f'Diabetes SVR - MSE: {mse_diab:.4f}, R2: {r2_diab:.4f}')
````
