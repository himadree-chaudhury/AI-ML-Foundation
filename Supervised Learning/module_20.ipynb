# Adaptive Boosting (Adaboost)
Adaboost, short for Adaptive Bloosting, is an ensemble learning technique that combines multiple weak classifiers to create a strong classifier. It was introduced by Yoav Freund and Robert Schapire in 1995. The main idea behind Adaboost is to focus on the instances that are hard to classify by giving them more weight in subsequent rounds of training.

### Why we need Adaboost?
In many real-world scenarios, individual classifiers may not perform well on their own, especially when dealing with complex datasets. Adaboost addresses this issue by combining multiple weak classifiers, which are only slightly better than random guessing, to create a more accurate and robust model. By iteratively adjusting the weights of misclassified instances, Adaboost effectively focuses on the difficult cases, leading to improved overall performance. In short, many weak models trained sequentially, each focusing on previous errors.

`Adaptive` means it changes focus during training based on performance
- Misclassified samples receive increased attention in later rounds
- Each new model tries to correct the mistakes of its predecessors

Core Principle: AdaBoost learns by fixing its past mistakes.


## Boosting
A single model may achieve decent overall accuracy but consistently struggle with difficult edge cases - the examples that matter most in real applications, here comes the boosting technique. Boosting is an ensemble technique that combines multiple weak learners to create a strong learner. The key idea is to train models sequentially, where each new model focuses on the errors made by the previous models. By giving more weight to misclassified instances, boosting aims to improve the overall performance of the ensemble. A weak learner is a deliberately simple model that is just slightly better than random guessing. They are fast to train, resistant to overfitting, and easy to combine. Their simplicity is a feature, not a flaw beacuse they need less computational resources and can be trained quickly on large datasets. When combined strategically, hundreds of weak learners create a model far stronger than any single complex model could be.

## Sequential Learning
Boosting relies on sequential learning, where each model is trained one after the other. Each new model focuses on the mistakes made by the previous models (dependency chain). This sequential approach allows the ensemble to adapt and improve over time, as each model learns from the errors of its predecessors. The final prediction is made by combining the outputs of all models, often through weighted voting or averaging, where models that perform better have a greater influence on the final decision.

## Sample Weighting
In boosting, sample weighting is a crucial technique used to emphasize the importance of certain instances in the training process. Initially, all instances are assigned equal weights. After each iteration, the weights of misclassified instances are increased, making them more influential in the training of the next model. This way, subsequent models focus more on the difficult cases that previous models struggled with. The weight adjustment is typically based on the error rate of the previous model, ensuring that harder-to-classify instances receive more attention in future rounds.

## Weighted Voting
In boosting, weighted voting is a method used to combine the predictions of multiple models in the ensemble. Each model's contribution to the final prediction is weighted based on its performance, with better-performing models receiving higher weights. This approach ensures that models that are more accurate have a greater influence on the final decision, while less accurate models contribute less. Weighted voting helps to improve the overall accuracy of the ensemble by leveraging the strengths of the individual models while minimizing the impact of their weaknesses. A key thing is that more learners does not always mean better performance, as adding too many weak learners can lead to overfitting.

## Adaboost Algorithm Steps
Core Idea of Boosting:
- Train a simple model (weak learner) on the data.
- Indentify the mistakes made by the model.
- Increase importance (weight) of misclassified instances.
- Train the next model focusing on these harder cases.
- Repeat steps 2-4 for several iterations.

1. Initialization of sample weights
    ```math
    w_i^{(1)} = \frac{1}{N}
    ```
    Where $N$ is the total number of training instances. This weight will be assigned to each instance initially.

2. After training the first weak learner, calculate its error rate:
    ```math
    \text{Weighted Error} = \epsilon_1 = \sum_{i=1}^{N} w_i \cdot || (y_i \neq h^{(1)}(x_i) ) ||
    ```
    Where $\epsilon_1$ is the weighted error of the first weak learner, $w_i$ is the weight of instance $i$, $y_i$ is the true label, and $h^{(1)}(x_i)$ is the predicted label by the weak learner $h^{(1)}$.

3. Compute the model weight:
    ```math
    \alpha_1 = \frac{1}{2} \ln\left(\frac{1 - \epsilon_1}{\epsilon_1}\right)
    ```
    Where $\alpha_1$ is the weight assigned to the first weak learner based on its performance.

4. Update the sample weights for the next iteration:
    ```math
    w_i^{new} = w_i^{(1)} \cdot e^{-\alpha_1 \cdot y_i \cdot h^{(1)}(x_i)}
    ```
    Where $w_i^{new}$ is the updated weight for instance $i$. The factor $y_i h^{(1)}(x_i)$ will be +1 if correctly classified and -1 if misclassified. Therefore:

    1. If instance $i$ is correctly classified:
    ```math
    w_i^{new} = w_i^{(1)} \cdot e^{-\alpha_1}
    ```
    2. If instance $i$ is misclassified:
    ```math
    w_i^{new} = w_i^{(1)} \cdot e^{\alpha_1}
    ```

5. Normalize the weights so that they sum to 1:
    ```math
    w_i^{(2)} = \frac{w_i^{new}}{\sum_{j=1}^{N} w_j^{new}}
    ```
    Where $w_i^{(2)}$ is the normalized weight for instance $i$ to be used in the second iteration. It will be seen that misclassified instances will have higher weights compared to correctly classified ones.

6. Repeat steps 2-5 for subsequent weak learners.

7. Final Prediction:
    The final prediction of the ensemble is made by combining the predictions of all weak learners, weighted by their respective model weights:
    ```math
    H(x) = \text{sign}\left(\sum_{m=1}^{M} \alpha_m \cdot h^{(m)}(x)\right)
    ```
    Where $H(x)$ is the final prediction, $M$ is the total number of weak learners, $\alpha_m$ is the weight of the $m$-th weak learner, and $h^{(m)}(x)$ is the prediction of the $m$-th weak learner. The sign function determines the final class label based on the weighted sum of predictions, it will output +1 if the sum is positive and -1 if negative.


##### Example
Consider a binary classification dataset: $D=\{(x_1,y_1)\}_{i=1}^5$ where $x_i$ are the features and $y_i \in \{-1, +1\}$ are the labels. For simplicity, let's assume we have 5 training samples:\
| Feature (x) | Label (y) |
|--------------|------------|
| $x_1$        | +1         |
| $x_2$        | +1         |
| $x_3$        | -1         |
| $x_4$        | -1         |
| $x_5$        | +1         |

- Initialize weights:
```math
w_i^{(1)} = \frac{1}{5} = 0.2\\
w_1 = w_2 = w_3 = w_4 = w_5 = 0.2
```

- After training the first weak learner, suppose it misclassifies $x_2$ and $x_4$. The weighted error is:
```math
\epsilon_1 = \sum_{i=1}^{5} w_i \cdot || (y_i \neq h^{(1)}(x_i) ) || = 0.2 + 0.2 = 0.4
```

- Compute model weight:
```math
\alpha_1 = \frac{1}{2} \ln\left(\frac{1 - 0.4}{0.4}\right) = \frac{1}{2} \ln(1.5) \approx 0.2027
```

- Update sample weights:
For correctly classified instances ($x_1$, $x_3$, $x_5$):
```math
w_i^{new} = 0.2 \cdot e^{-0.2027} \approx 0.1637
```
For misclassified instances ($x_2$, $x_4$):
```math
w_i^{new} = 0.2 \cdot e^{0.2027} \approx 0.2443
```

- Normalize weights:
```math
\text{Sum of new weights} = 3 \cdot 0.1637 + 2 \cdot 0.2443 = 0.4911 + 0.4886 = 0.9797
```
```math
w_1^{(2)} = \frac{0.1637}{0.9797} \approx 0.1671\\
```
```math
w_2^{(2)} = \frac{0.2443}{0.9797} \approx 0.2493\\
```
```math
w_3^{(2)} = \frac{0.1637}{0.9797} \approx 0.1671\\
```
```math
w_4^{(2)} = \frac{0.2443}{0.9797} \approx 0.2493\\
```
```math
w_5^{(2)} = \frac{0.1637}{0.9797} \approx 0.1671
```
- The updated weights for the next iteration are:
```math
w_1^{(2)} \approx 0.1671, w_2^{(2)} \approx 0.2493, w_3^{(2)} \approx 0.1671, w_4^{(2)} \approx 0.2493, w_5^{(2)} \approx 0.1671
```
As we can see, the weights for misclassified instances ($x_2$ and $x_4$) have increased, while the weights for correctly classified instances have decreased.

- Repeat the process for subsequent weak learners, adjusting weights and combining predictions until the desired number of weak learners is reached.


#### Bagging vs Boosting
| Aspect               | Bagging                              | Boosting                             |
|----------------------|-------------------------------------|-------------------------------------|
| Model Training       | Parallel                            | Sequential                          |
| Focus                | Reduces variance                    | Reduces bias                        |
| Instance Weighting   | Equal weights for all instances     | Higher weights for misclassified instances |
| Model Combination    | Majority voting or averaging        | Weighted voting based on model performance |


````python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier

X, y = load_breast_cancer(return_X_y=True)  # retuern_X_y for getting data and labels directly. It split the target and features into X and y.

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

base_learner = DecisionTreeClassifier(max_depth=1, random_state=42) # the baseline/weak model selected before boosting via testing

ada = AdaBoostClassifier(
    estimator = base_learner,
    n_estimators=20,
    learning_rate=0.3,
    random_state=42
)

ada.fit(X_train, y_train)
````

````python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred_ada = ada.predict(X_test)
print('AdaBoost Accuracy: ', accuracy_score(y_test, y_pred_ada))
print(classification_report(y_test, y_pred_ada))
print('Confusion Matrix:\n', confusion_matrix(y_test, y_pred_ada))
````

````python
# Hyperparameter Tuning with Grid Search CV

from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': list(range(20, 200,2)),
    'learning_rate': [i/100 for i in range(1, 10)],
    'estimator__max_depth': [1, 2]
}

grid = GridSearchCV(
    estimator=AdaBoostClassifier(
        estimator=base_learner,
        random_state=42
    ),
    param_grid=param_grid,
    scoring='accuracy',
    cv=5,
    n_jobs=-1
)

grid.fit(X_train, y_train)
````

````python
best_model = grid.best_estimator_
print('Best Parameters: ', grid.best_params_)
````

````python
y_pred_grid = best_model.predict(X_test)
print('Tuned AdaBoost Accuracy: ', accuracy_score(y_test, y_pred_grid))
print(classification_report(y_test, y_pred_grid))
print('Confusion Matrix:\n', confusion_matrix(y_test, y_pred_grid))
````
