# Question 01: [ Marks 10 ]

What is a perceptron?
Explain its three main components and their roles.

## Write Answer 01: 
Perceptron is a type of artificial neural networks(ANNs) that is used for binary classification tasks in deep learning.

Perceptron consists of three main components: input features, weights, and bias. The input features are the data points that are to train the model. The weights determine the importance of each input feature in making a prediction. The bias is an additional parameter that allows the decision boundary to be shifted. The perceptron uses these components to calculate a weighted sum of the input features and applies an activation function to produce a binary output (0 or 1) based on whether the weighted sum exceeds a certain threshold.

# Question 02: [ Marks 10 ]

What is a decision boundary in a perceptron?
Explain how the equation
w₁x₁ + w₂x₂ + b = 0
represents a decision boundary geometrically.

## Write Answer 02:
The decision boundary in a perceptron is a line that separates the feature space into two regions, each corresponding to a different class. It is defined by the equation $w_1x_1 + w_2x_2 + b = 0$, where $w_1$ and $w_2$ are the weights associated with the input features $x_1$ and $x_2$, and $b$ is the bias term. If the prediction is greater than or equal to 0, the output is class 1; otherwise, it is class 0. Geometrically, this equation represents a straight line in a two-dimensional feature space. The weights determine the slope of the line, while the bias determines its position. 

# Question 03: [ Marks 15 ]

Why is the perceptron called a linear classifier?
Explain your answer using the concept of linear separability and logical gate examples such as AND, OR, and XOR.

## Write Answer 03:
The perceptron is called a linear classifier because it makes its predictions based on a linear combination of the input features. The decision boundary created by a perceptron is a straight line that separates the feature space into two regions corresponding to different classes. The perceptron can only classify data that is linearly separable, meaning that there exists a straight line that can separate the classes without any misclassification. 

In the case of logical gates, the AND and OR gates are linearly separable because we can draw a straight line that separates the output classes (0 and 1) based on the input features. However, the XOR gate is not linearly separable because there is no straight line that can separate the output classes without misclassification. Therefore, while the perceptron can model AND and OR gates, it cannot model the XOR gate due to its non-linear nature.

# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:
The perceptron weight update rule is given by:\
$w = w + learning\_rate \times (y - y_{pred}) \times x$\
$b = b + learning\_rate \times (y - y_{pred})$

Where:
- $w$ is the weight vector
- $b$ is the bias
- $learning\_rate$ is a hyperparameter that controls the step size of the update
- $y$ is the actual class label
- $y_{pred}$ is the predicted class label

Intuitively, the weight update rule adjusts the weights and bias in the direction that reduces the error between the predicted output and the actual output. If the prediction is correct ($y_{pred} = y$), then there is no error, and the weights do not need to be updated. However, if the prediction is incorrect ($y_{pred} \neq y$), then the weights are updated to reduce the error for that specific data point. This is why the weights are updated only when the prediction is incorrect, as it allows the model to learn from its mistakes and improve its performance over time.

# Question 05: [ Marks 10 ]

Why does a single-layer perceptron fail to solve the XOR problem?
What does this limitation tell us about the need for multilayer neural networks?

## Write Answer 05:
A single-layer perceptron fails to solve the XOR problem because the XOR function is not linearly separable. In other words, there is no straight line that can separate the output classes (0 and 1) based on the input features for the XOR function. The XOR function outputs 1 only when the inputs are different (0,1 or 1,0) and outputs 0 when the inputs are the same (0,0 or 1,1). This creates a situation where the data points cannot be separated by a single linear decision boundary.

This limitation of the single-layer perceptron highlights the need for multilayer neural networks, such as the multi-layer perceptron (MLP), which can learn non-linear decision boundaries. This is why multilayer neural networks are essential for solving non linearly separable classification problems in deep learning.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:
The perceptron Learning Rule, also known as the weight adjustment method, is a supervised learning algorithm used to train a perceptron. It involves updating the weights and bias of the perceptron based on the error between the predicted output and the actual output for each training example. The update is done using the formula:
$w = w + learning\_rate \times (y - y_{pred}) \times x$\
$b = b + learning\_rate \times (y - y_{pred}) $

Where:
- $w$ is the weight vector
- $b$ is the bias
- $learning\_rate$ is a hyperparameter that controls the step size of the update
- $y$ is the actual class label
- $y_{pred}$ is the predicted class label

Adjusting one weight can affect the classification of other points because the decision boundary of the perceptron is determined by the weights and bias. When a weight is adjusted, it changes the slope and position of the decision boundary. This can lead to a situation where some data points that were previously classified correctly may now be misclassified, while others that were misclassified may now be classified correctly. Therefore, the adjustment of one weight can have a ripple effect on the classification of multiple data points, highlighting the interconnected nature of the weights in determining the overall performance of the perceptron.

# Question 07: [ Marks 20 ]

What are the main limitations of a single-layer perceptron?
How does the idea of using more than one layer (MLP) help overcome these limitations at a high level accoriding to you?



## Write Answer 07:
The main limitations of a single-layer perceptron include:
1. Inability to solve non-linearly separable problems: A single-layer perceptron can only classify data that is linearly separable. It fails to classify data points that cannot be separated by a straight line, such as the XOR problem.
2. Limited representational power: A single-layer perceptron can only model linear decision boundaries, which limits its ability to capture complex relationships between input features and output classes.
3. Sensitivity to outliers: A single-layer perceptron can be sensitive to outliers in the training data, which can lead to poor generalization performance.

The idea of using more than one layer, as in a multi-layer perceptron (MLP), helps overcome these limitations by allowing the model to learn non-linear decision boundaries. By adding hidden layers and non-linear activation functions, MLPs can capture complex relationships between input features and output classes that a single-layer perceptron cannot. This enables MLPs to solve problems that are not linearly separable, such as the XOR problem, and to model more complex patterns in the data, leading to improved performance on a wider range of classification tasks in deep learning.
