# Multi Layer Perceptron (MLP)
A Multi Layer Perceptron (MLP) is a type of feedforward artificial neural network that consists of multiple layers of nodes, where each layer is fully connected to the next one. The MLP is designed to model complex relationships between inputs and outputs by learning from data. Here all the input is fed into all the nodes in the next layer, and this process continues until the output layer is reached. The MLP can be used for various tasks such as classification, regression, and function approximation. It typically uses activation functions like ReLU or sigmoid to introduce non-linearity into the model, allowing it to learn more complex patterns in the data.

## Neural Layer
A neural layer is a fundamental component of a neural network, consisting of a group of neurons(node) that process input data and produce output. Each neuron in the layer receives input from the previous layer, applies a weighted sum to the inputs, and then passes the result through an activation function to produce an output. The weights are adjusted during training to minimize the error between the predicted output and the actual output. Input layers receive the initial data, hidden layers perform intermediate processing, and output layers produce the final predictions. Neural layers are essential for enabling the network to learn and model complex relationships in the data.

### Parameters of a Neural Layer
The parameters of a neural layer typically include:
1. Weights: These are the coefficients that determine the strength of the connection between neurons in different layers. They are adjusted during training to minimize the error in predictions. The representation of weights can be denoted as: $w^k_{ij}$; Where $w$ represents the weight for the connection, $k$ represents the layer number that the weight goes to, $i$ represents the node the input is coming from, and $j$ represents the node the input is going to in the next layer.
2. Biases: These are additional parameters that allow the activation function to be shifted to the left or right, which can help the model fit the data better. Bias is often represented as: $b_{ij}$; Where $b$ represents the bias for the connection and $i$ represents the current layer, while $j$ represents the neuron in the current layer.
3. Activation Function: This is a function applied to the output of each neuron to introduce non-linearity into the model. Common activation functions include ReLU, sigmoid, and tanh.
4. Number of Neurons: This refers to the number of nodes in the layer, which can affect the capacity of the model to learn from data. More neurons can capture more complex patterns, but may also lead to overfitting if not properly regularized.
5. Output: This is the result produced by the each node in the layer after applying the weights, biases, and activation function. The output of one layer serves as the input for the next layer in the network. Output can be represented as: $O_{ij}$; Where $O$ represents the output of the neuron, $i$ represents the current layer, and $j$ represents the neuron in the current layer.

## Forward Propagation
Forward propagation is the process by which input data is passed through a neural network to generate an output. During forward propagation, the input data is fed into the input layer, and then it is processed through each subsequent layer of the network. At each layer, the input is transformed by applying weights, biases, and activation functions to produce an output that serves as the input for the next layer. This process continues until the output layer is reached, where the final prediction or output is generated. Forward propagation is essential for making predictions with a trained neural network and is also used during the training phase to compute the loss and update the weights through backpropagation.

The formula for forward propagation can be expressed as follows:
```math
O = \sigma(W^T \cdot X + b) 
```
Where:
- $O$ is the output of the neuron.
- $\sigma$ is the activation function.
- $W^T$ is the transpose of the weight matrix.
- $X$ is the input vector.
- $b$ is the bias term.

### Example
Let a data have 4 features ${x_1, x_2, x_3, x_4}$. The first hidden layer has 3 neurons $O_{11}, O_{12}, O_{13}$ and the second hidden layer has 2 neurons $O_{21}, O_{22}$. The output layer has 1 neuron $O_{31}$. As the input layer receives the data, each neuron in the first hidden layer receives 4 inputs, the second hidden layer receives 3 inputs from the first hidden layer, and the output layer receives 2 inputs from the second hidden layer. So the number of weights in the first hidden layer is $4 \times 3 = 12$ ${w^1_{11}, w^1_{12}, w^1_{13}, w^1_{14}, w^1_{21}, w^1_{22}, w^1_{23}, w^1_{24}, w^1_{31}, w^1_{32}, w^1_{33}, w^1_{34}, w^1_{41}, w^1_{42}, w^1_{43}, w^1_{44}}$, in the second hidden layer is $3 \times 2 = 6$ ${w^2_{11}, w^2_{12}, w^2_{21}, w^2_{22}, w^2_{31}, w^2_{32}}$, and in the output layer is $2 \times 1 = 2$ ${w^3_{11}, w^3_{21}}$. The total number of weights in the MLP is $12 + 6 + 2 = 20$. For biases, each neuron in the hidden layers and output layer has one bias, so the total number of biases is $3 + 2 + 1 = 6$ ${b_{11}, b_{12}, b_{13}, b_{21}, b_{22}, b_{31}}$. Therefore, the total number of parameters in this MLP is $20$ (weights) + $6$ (biases) = $26$ parameters.

Now, let's calculate the output of the first hidden layer using forward propagation. The weight matrix for the first hidden layer can be represented as:
```math
W^1 = \begin{bmatrix}w^1_{11} & w^1_{12} & w^1_{13} \\ w^1_{21} & w^1_{22} & w^1_{23} \\ w^1_{31} & w^1_{32} & w^1_{33} \\ w^1_{41} & w^1_{42} & w^1_{43}\end{bmatrix}
```
Transpose of the weight matrix is:
```math
(W^1)^T = \begin{bmatrix}w^1_{11} & w^1_{21} & w^1_{31} & w^1_{41} \\ w^1_{12} & w^1_{22} & w^1_{32} & w^1_{42} \\ w^1_{13} & w^1_{23} & w^1_{33} & w^1_{43}\end{bmatrix}
```
The input vector is:
```math
X = \begin{bmatrix}x_1 \\ x_2 \\ x_3 \\ x_4\end{bmatrix}
```
The bias vector for the first hidden layer is:
```math
b^1 = \begin{bmatrix}b_{11} \\ b_{12} \\ b_{13}\end{bmatrix}
```

Now, we can calculate the output of the first hidden layer using the forward propagation formula:
```math
O^1 = \sigma((W^1)^T \cdot X + b^1)
```
Where $O^1$ is the output of the first hidden layer, $\sigma$ is the activation function applied element-wise to the result of the matrix multiplication and addition. 
```math
O^1 = \sigma\left(\begin{bmatrix}w^1_{11} & w^1_{21} & w^1_{31} & w^1_{41} \\ w^1_{12} & w^1_{22} & w^1_{32} & w^1_{42} \\ w^1_{13} & w^1_{23} & w^1_{33} & w^1_{43}\end{bmatrix} \cdot \begin{bmatrix}x_1 \\ x_2 \\ x_3 \\ x_4\end{bmatrix} + \begin{bmatrix}b_{11} \\ b_{12} \\ b_{13}\end{bmatrix}\right)
```
```math
O^1 = \sigma\left(\begin{bmatrix}w^1_{11}x_1 + w^1_{21}x_2 + w^1_{31}x_3 + w^1_{41}x_4 + b_{11} \\ w^1_{12}x_1 + w^1_{22}x_2 + w^1_{32}x_3 + w^1_{42}x_4 + b_{12} \\ w^1_{13}x_1 + w^1_{23}x_2 + w^1_{33}x_3 + w^1_{43}x_4 + b_{13}\end{bmatrix}\right)
```
```math
O^1 = \begin{bmatrix}(O^1_{11}) \\ (O^1_{12}) \\ (O^1_{13})\end{bmatrix} = \begin{bmatrix}\sigma(w^1_{11}x_1 + w^1_{21}x_2 + w^1_{31}x_3 + w^1_{41}x_4 + b_{11}) \\ \sigma(w^1_{12}x_1 + w^1_{22}x_2 + w^1_{32}x_3 + w^1_{42}x_4 + b_{12}) \\ \sigma(w^1_{13}x_1 + w^1_{23}x_2 + w^1_{33}x_3 + w^1_{43}x_4 + b_{13})\end{bmatrix}
```

This $O^{11}$, $O^{12}$, and $O^{13}$ are the outputs of the three neurons in the first hidden layer, which will be used as inputs for the second hidden layer. The same process can be repeated to calculate the outputs of the second hidden layer and the output layer using their respective weight matrices, bias vectors, and activation functions.
```math
O^2 = \sigma((W^2)^T \cdot O^1 + b^2)
```
```math
O^2 = \sigma\left(\begin{bmatrix}w^2_{11} & w^2_{21} & w^2_{31} \\ w^2_{12} & w^2_{22} & w^2_{32}\end{bmatrix} \cdot \begin{bmatrix}(O^1_{11}) \\ (O^1_{12}) \\ (O^1_{13})\end{bmatrix} + \begin{bmatrix}b_{21} \\ b_{22}\end{bmatrix}\right)
```
```math
O^2 = \sigma\left(\begin{bmatrix}w^2_{11}O^1_{11} + w^2_{21}O^1_{12} + w^2_{31}O^1_{13} + b_{21} \\ w^2_{12}O^1_{11} + w^2_{22}O^1_{12} + w^2_{32}O^1_{13} + b_{22}\end{bmatrix}\right)
```
```math
O^2 = \begin{bmatrix}(O^2_{21}) \\ (O^2_{22})\end{bmatrix} = \begin{bmatrix}\sigma(w^2_{11}O^1_{11} + w^2_{21}O^1_{12} + w^2_{31}O^1_{13} + b_{21}) \\ \sigma(w^2_{12}O^1_{11} + w^2_{22}O^1_{12} + w^2_{32}O^1_{13} + b_{22})\end{bmatrix}
```
Finally, the output of the output layer can be calculated as:
```math
O^3 = \sigma((W^3)^T \cdot O^2 + b^3)
```
```math
O^3 = \sigma\left(\begin{bmatrix}w^3_{11} & w^3_{21}\end{bmatrix} \cdot \begin{bmatrix}(O^2_{21}) \\ (O^2_{22})\end{bmatrix} + \begin{bmatrix}b_{31}\end{bmatrix}\right)
```
```math
O^3 = \sigma\left(w^3_{11}O^2_{21} + w^3_{21}O^2_{22} + b_{31}\right)
```
This $O^3$ is the final output of the MLP, which can be used for making predictions or further processing depending on the task at hand.

In some research papers, the output of each layer is represented as $\alpha[i]=\sigma(\alpha[i-1] \cdot W^i + b^i)$, where $\alpha[i]$ is the output of the $i^{th}$ layer, $\alpha[i-1]$ is the output of the previous layer, $W^i$ is the weight matrix for the $i^{th}$ layer, and $b^i$ is the bias vector for the $i^{th}$ layer. This notation emphasizes the recursive nature of forward propagation through the layers of the network.
