# Question 05: [ Marks 10 ]

Create two tensors:
w = 2.0 (requires_grad=True) <br>
b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:

````python
import torch
w= torch.tensor(2.0, requires_grad=True)
b= torch.tensor(1.0, requires_grad=True)
x= torch.tensor(3.0)
y = torch.tensor(1.0)

z= w*x + b
y_pred = torch.sigmoid(z)
y_clamped = torch.clamp(y_pred, min=1e-7, max=1-1e-7)
BCE_loss = - (y * torch.log(y_clamped) + (1 - y) * torch.log(1 - y_clamped))
BCE_loss.backward()

print(f"Predicted value (y_pred): {y_pred.item():.4f}")
print(f"Binary Cross-Entropy Loss: {BCE_loss.item():.4f}")
print(f"Gradient of w: {w.grad:.4f}")
print(f"Gradient of b: {b.grad:.4f}")
````

# Question 07: [ Marks 20 ]


Create tensor w = 2.0 (requires_grad=True). <br>

Compute two different losses sequentially and call backward() twice. <br>
Print w.grad. <br>

Do not clear gradients. <br>
Explain what you observe. <br>

Write full code.


## Write Answer 07:



````python
w= torch.tensor(2.0, requires_grad=True)

loss1 = w**2 + 3*w + 5
loss1.backward()

print(f"Gradient after 1st backward pass: {w.grad}")

loss2 = w**2
loss2.backward()

print(f"Gradient after 2nd backward pass: {w.grad}")
````

After computing the first loss and calling backward(), w.grad will contain the gradient of the first loss with respect to w. $dloss1/dw = 2w+3 = 2*2+3=7$ and shows it when I print w.grad. When I compute the second loss and call backward() again, the new gradient will be $dloss2/dw = w^2 = 2^2 = 4$. Since I did not clear the gradients, w.grad will now contain the sum of the gradients from both losses, which is $7 + 4 = 11$. This shows that when I call backward() multiple times without clearing gradients, the gradients accumulate in the .grad attribute.
