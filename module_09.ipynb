# Backpropagation
Backpropagation is a fundamental algorithm used in training artificial neural networks. It is a supervised learning technique that allows the network to adjust its weights and biases based on the error between the predicted output and the actual output. The main idea behind backpropagation is to propagate the error backward through the network, allowing each layer to update its parameters accordingly.The backpropagation algorithm consists of two main phases: the forward pass and the backward pass.\
1. Forward Pass: In this phase, the input data is passed through the network layer by layer, and the output is computed. The activations of each layer are stored for use in the backward pass.
2. Backward Pass: In this phase, the error is calculated at the output layer by comparing the predicted output with the actual output. The error is then propagated backward through the network, and the weights and biases are updated using gradient descent. The gradients of the loss function with respect to each parameter are computed, and the parameters are adjusted in the direction that minimizes the loss.

The main concept is, when the model makes a prediction, it calculates the error (the difference between the predicted output and the actual output). If the error is large, the model needs to adjust its parameters (weights and biases) to reduce the error. Backpropagation updates the weights and biases in such a way that the error is minimized, allowing the model to learn from its mistakes and improve its predictions over time. The core ides is to enable the model to learn from its mistakes and improve its predictions over time.

### Backpropagation in Steps
1. Initialization: The weights and biases of the neural network are initialized, often with small random values.
2. Prediction: The input data is fed through the network to generate a predicted output(Forward Pass).
3. Error Calculation: The error is calculated by comparing the predicted output with the actual output using a loss function (e.g., Mean Squared Error, Cross-Entropy Loss).
4. Backward Pass: The error is propagated backward through the network, and the gradients of the loss function with respect to each weight and bias are computed.
5. Parameter Update: The weights and biases are updated using an optimization algorithm (e.g., Stochastic Gradient Descent) to minimize the loss.

#### Example of Backpropagation
Let's consider a simple neural network with $2$ input neurons, $1$ hidden layer with $2$ neurons, and $1$ output neuron. The inputs are $x_1$ and $x_2$, the weights are $w_{11}^{1}$, $w_{12}^{1}$, $w_{21}^{1}$, and $w_{22}^{1}$, biases are $b_{11}$ and $b_{12}$, and the output are $O_{11}$ and $O_{12}$ for the 1st hidden layer. The weights for the output layer are $w_{11}^{2}$ and $w_{21}^{2}$, and the bias is $b_{21}$. The output of the network is $O_{21}$.
1. Forward Pass:
   - Compute the activations of the hidden layer:
     - $O_{11} = f(w_{11}^{1} \cdot x_1 + w_{12}^{1} \cdot x_2 + b_{11})$
     - $O_{12} = f(w_{21}^{1} \cdot x_1 + w_{22}^{1} \cdot x_2 + b_{12})$
   - Compute the output of the network:
     - $O_{21} = f(w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21})$

   Here, $f$ is the activation function, which could be a sigmoid, ReLU, or any other non-linear function.
2. Error Calculation:
   - Calculate the error using a loss function, for example, Mean Squared Error:
     - $E = \frac{1}{2}(O_{21} - y)^2$, where $y$ is the actual output.
3. Backward Pass:
   - Compute the gradients for the output layer:
     - $\frac{\partial E}{\partial w_{11}^{2}} = \frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial w_{11}^{2}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial }{\partial w_{11}^{2}} ({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) = (O_{21} - y) \cdot f'(O_{21}) \cdot O_{11}$
     - $\frac{\partial E}{\partial w_{21}^{2}} = \frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial w_{21}^{2}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial }{\partial w_{21}^{2}} ({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) = (O_{21} - y) \cdot f'(O_{21}) \cdot O_{12}$
     - $\frac{\partial E}{\partial b_{21}} = \frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial b_{21}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial }{\partial b_{21}} ({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) = (O_{21} - y) \cdot f'(O_{21})$
   - Compute the gradients for the hidden layer:
     - $\frac{\partial E}{\partial w_{11}^{1}} = \frac{\partial E}{\partial O_{11}} \cdot \frac{\partial O_{11}}{\partial w_{11}^{1}} = (\frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial O_{11}}) \cdot \frac{\partial O_{11}}{\partial w_{11}^{1}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial}{\partial O_{11}}({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) \cdot \frac{\partial }{\partial w_{11}^{1}} ({w_{11}^{1} \cdot x_1 + w_{12}^{1} \cdot x_2 + b_{11}}) = (O_{21} - y) \cdot w_{11}^{2} \cdot f'(O_{11}) \cdot x_1$
     - $\frac{\partial E}{\partial w_{12}^{1}} = \frac{\partial E}{\partial O_{12}} \cdot \frac{\partial O_{12}}{\partial w_{12}^{1}} = (\frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial O_{12}}) \cdot \frac{\partial O_{12}}{\partial w_{12}^{1}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial}{\partial O_{12}}({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) \cdot \frac{\partial }{\partial w_{12}^{1}} ({w_{21}^{1} \cdot x_1 + w_{22}^{1} \cdot x_2 + b_{12}}) = (O_{21} - y) \cdot w_{21}^{2} \cdot f'(O_{12}) \cdot x_2$
     - $\frac{\partial E}{\partial b_{11}} = \frac{\partial E}{\partial O_{11}} \cdot \frac{\partial O_{11}}{\partial b_{11}} = (\frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial O_{11}}) \cdot \frac{\partial O_{11}}{\partial b_{11}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial}{\partial O_{11}}({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) \cdot \frac{\partial }{\partial b_{11}} ({w_{11}^{1} \cdot x_1 + w_{12}^{1} \cdot x_2 + b_{11}}) = (O_{21} - y) \cdot w_{11}^{2} \cdot f'(O_{11}) $
     - $\frac{\partial E}{\partial w_{21}^{1}} = \frac{\partial E}{\partial O_{12}} \cdot \frac{\partial O_{12}}{\partial w_{21}^{1}} = (\frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial O_{12}}) \cdot \frac{\partial O_{12}}{\partial w_{21}^{1}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial}{\partial O_{12}}({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) \cdot \frac{\partial }{\partial w_{21}^{1}} ({w_{21}^{1} \cdot x_1 + w_{22}^{1} \cdot x_2 + b_{12}}) = (O_{21} - y) \cdot w_{21}^{2}  \cdot f'(O_{12}) \cdot x_1$
     - $\frac{\partial E}{\partial w_{22}^{1}} = \frac{\partial E}{\partial O_{12}} \cdot \frac{\partial O_{12}}{\partial w_{22}^{1}} = (\frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial O_{12}}) \cdot \frac{\partial O_{12}}{\partial w_{22}^{1}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial}{\partial O_{12}}({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) \cdot \frac{\partial }{\partial w_{22}^{1}} ({w_{21}^{1} \cdot x_1 + w_{22}^{1} \cdot x_2 + b_{12}}) = (O_{21} - y) \cdot w_{21}^{2}  \cdot f'(O_{12}) \cdot x_2$
     - $\frac{\partial E}{\partial b_{12}} = \frac{\partial E}{\partial O_{12}} \cdot \frac{\partial O_{12}}{\partial b_{12}} = (\frac{\partial E}{\partial O_{21}} \cdot \frac{\partial O_{21}}{\partial O_{12}}) \cdot \frac{\partial O_{12}}{\partial b_{12}} = \frac{\partial}{\partial O_{21}} {\frac{1}{2} {(O_{21}-y)^2}} \cdot \frac{\partial}{\partial O_{12}}({w_{11}^{2} \cdot O_{11} + w_{21}^{2} \cdot O_{12} + b_{21}}) \cdot \frac{\partial }{\partial b_{12}} ({w_{21}^{1} \cdot x_1 + w_{22}^{1} \cdot x_2 + b_{12}}) = (O_{21} - y) \cdot w_{21}^{2} \cdot f'(O_{12})$

    Here, $f'$ is the derivative of the activation function, which is used to compute the gradients during backpropagation.
4. Parameter Update:
   - Update the weights and biases using an optimization algorithm, for example, Stochastic Gradient Descent:
     - $w_{11}^{2} = w_{11}^{2} - \alpha \cdot \frac{\partial E}{\partial w_{11}^{2}}$
     - $w_{21}^{2} = w_{21}^{2} - \alpha \cdot \frac{\partial E}{\partial w_{21}^{2}}$
     - $b_{21} = b_{21} - \alpha \cdot \frac{\partial E}{\partial b_{21}}$
     - $w_{11}^{1} = w_{11}^{1} - \alpha \cdot \frac{\partial E}{\partial w_{11}^{1}}$
     - $w_{12}^{1} = w_{12}^{1} - \alpha \cdot \frac{\partial E}{\partial w_{12}^{1}}$
     - $b_{11} = b_{11} - \alpha \cdot \frac{\partial E}{\partial b_{11}}$
     - $w_{21}^{1} = w_{21}^{1} - \alpha \cdot \frac{\partial E}{\partial w_{21}^{1}}$
     - $w_{22}^{1} = w_{22}^{1} - \alpha \cdot \frac{\partial E}{\partial w_{22}^{1}}$
     - $b_{12} = b_{12} - \alpha \cdot \frac{\partial E}{\partial b_{12}}$

    Where $\alpha$ is the learning rate, which controls the step size of the parameter updates. 

This process is repeated for multiple iterations (epochs) until the model converges to a solution that minimizes the loss function.
